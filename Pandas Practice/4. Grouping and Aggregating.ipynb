# **Grouping and Aggregating**

```python
import numpy as np
import pandas as pd
```

## **Aggregation**
-used to get a summary of columns in our dataset

```python
df = pd.DataFrame({'Maths':[9, 4, 8, 9],
                   'English':[8, 10, 7, 6],
                   'Science':[7, 6, 8, 5]})
print(df)
```

```python
#Sum of values in each column
print(df.sum())
```

```python
#Computing mean of each column
print(df.mean())
```

```python
#Generating all important statistics in one go
print(df.describe())
```

```python
#applying multiple aggregate function at a time
print(df.agg(['sum', 'min', 'mean']))
```

## **Grouping**

```python
data = {
    'Item': ['Cake', 'Cake', 'Bread', 'Pastry', 'Cake'],
    'Flavor': ['Chocolate', 'Vanilla', 'Whole Wheat', 'Strawberry', 'Chocolate'],
    'Price': [250, 220, 80, 120, 250]
}

df = pd.DataFrame(data)
print(df)
```

```python
#grouping data by a column "Item"
print(df.groupby('Item'))
```

```python
#getting sum of prices each items
print(df.groupby('Item')['Price'].sum())
```

```python
#grouping by multiple columns
print(df.groupby(['Item', 'Flavor'])['Price'].sum())
```

```python
#selecting a group
print(df.groupby('Item').get_group('Cake'))
```
