## **Saving a Pandas Dataframe as a File**

```python
#Creating a DataFrame
data = [
    {'Name': 'John', 'Age': 28, 'City': 'New York', 'Salary': 65000 },
    {'Name': 'Anna', 'Age': 34, 'City': 'Paris', 'Salary': 70000  },
    {'Name': 'Peter', 'Age': 29, 'City': 'Berlin', 'Salary': 62000 },
]
df = pd.DataFrame(data)
print(df)
```

```python
#Exporting DataFrame to CSV
df.to_csv('file1.csv')
```

```python
#Exporting DataFrame to CSV without index columns
df.to_csv('file1.csv', index = False)
```

```python
#Exporting only selected columns from DataFrame to CSV 
df.to_csv("file1.csv", columns = ['Name'])
```

```python
#Exporting DataFrame to CSV without header row
df.to_csv('file1.csv', header = False)
```

```python
#Exporting DataFrame to JSON as records(Each row becomes a separate JSON object)
df.to_json('file.json',orient='records')
```
