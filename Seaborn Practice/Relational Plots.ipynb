```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## **Loading Datasets**

```python
#veiwing all the names of example datasets
sns.get_dataset_names()
```

```python
#importing the dataset
df=sns.load_dataset('tips')
df
```

## **Scatter Plot**

```python
#simple scatter plot showing relation between total bills and tip
sns.relplot(x='total_bill', y='tip', data=df)
plt.show()
```

```python
#adding color(hue) based on sex
sns.relplot(x='total_bill', y='tip', data=df, hue='sex')
plt.show()
```

```python
#removing the legend
sns.relplot(x='total_bill', y='tip', data=df, hue='sex', legend=False)
plt.show()
```

```python
#changinng the size of dots wrt a variable
sns.relplot(data=df, x="total_bill", y="tip", size="size", hue='sex')
plt.show()
```

## **Line Plot**

```python
df = sns.load_dataset("iris")
df
```

```python
#simple line plot showing relation between sepal_length and sepal_width
sns.lineplot(x="sepal_length", y="sepal_width", data=df)
plt.show()
```

Note: More complex datasets will have multiple measurements for the same value of the x variable. The default behavior in seaborn is to aggregate the multiple measurements at each x value by plotting the mean and the 95% confidence interval around the mean:

```python
#lineplots based on species
sns.lineplot(x="sepal_length", y="sepal_width", data=df, hue='species')
plt.show()
```

## **Showing multiple relationships with facets**
means that you make multiple axes and plot subsets of the data on each axes

```python
#importing the dataset
df=sns.load_dataset('tips')
df
```

```python
#scatter plots showing relation between total bill and tip during different times
sns.relplot(data=df,x="total_bill", y="tip", hue="smoker", col="time")
plt.show()
```
